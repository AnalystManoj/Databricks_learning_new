```python
#tranforming sales_orders table 
from pyspark.sql.functions import *

# converting into dataframe
df = spark.table("databricks_simulated_retail_customer_data.v01.sales_orders")

# tranformations
df = df.withColumn(
    "order_date",
    col("order_datetime").cast("timestamp").cast("date")
).\
withColumn("order_datetime",col("order_datetime").cast("timestamp"))
```

```python
%sql
-- CREATE CATALOG my_retail_catalog;
-- SHOW CATALOGS;
-- CREATE SCHEMA my_retail_catalog.v01;
```

```python
# persisting transformed data
df.write \
  .format("delta") \
  .mode("overwrite") \
  .saveAsTable("my_retail_catalog.v01.sales_orders_transformed")
```
